```python
#Generalisation of the code so that we can use that for multiple book

def partition_function (name_of_book, number_of_partition):

  #get the book from the gutemberg collection
  text = nltk.corpus.gutenberg.raw(name_of_book)

  #first we tockenize the text
  tokenized_sents=nltk.sent_tokenize(text)

  #number of token in each partiion
  token_per_partition = int(len(tokenized_sents)/number_of_partition)

  #initialse the partition list of 200 partition
  partitions = []

  #loop throufg the tockenzed lsit and get the number of token needed for each partiton until we reach the number of partition given
  for i in range(number_of_partition):
    start = i * token_per_partition
    end = start + token_per_partition
    partitions.append(tokenized_sents[start:end])

  #we will append the name of the book these partition come from
  labeled_par = [(doc, name_of_book) for doc in partitions]

  return labeled_par
```

```python
#Apply the fonction to the three books
books_name = ['chesterton-thursday.txt', 'edgeworth-parents.txt', 'melville-moby_dick.txt']

output = []

for book in books_name:
  output.extend(partition_function(book, 200))

#shuffle the output
random.shuffle(output)
```

```python
#serialize the output with panda
import pandas as pd

rows = []
for i, (partition, book) in enumerate(output):
    rows.append({
        "book": book,
        "partition_id": i,
        "text": " ".join(partition)
    })

df = pd.DataFrame(rows)
```
